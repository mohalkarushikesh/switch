Python Dictionaries

A Python dictionary (dict) is a built-in data structure that stores data in key-value pairs, allowing for incredibly fast data retrieval. Dictionaries are mutable (changeable) and ordered (they maintain the order in which items are inserted).

Basic Rules
Keys: Must be unique and immutable data types (e.g., strings, numbers, or tuples).
Values: Can be any data type (e.g., lists, strings, integers, or even other dictionaries) and can repeat.
Core Operations & Examples
1. Creating a Dictionary

You can define a dictionary using curly braces {} or the dict() constructor.

# Using curly braces
user = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# Using the dict() constructor
user_alt = dict(name="Alice", age=25, city="New York")

2. Accessing Values

You can fetch a value using its key inside square brackets, or safely via the .get() method.

# Using square brackets (Raises KeyError if key is missing)
print(user["name"])  # Output: Alice

# Using .get() (Returns None or a default value if key is missing)
print(user.get("email"))        # Output: None
print(user.get("email", "N/A")) # Output: N/A

3. Adding and Modifying Items

Assign a value to a key directly using the = operator.

# Update an existing key
user["age"] = 26

# Add a new key-value pair
user["email"] = "alice@example.com"

4. Removing Items

You can use del, .pop(), or .clear() depending on your goal.

# Remove a specific key and return its value
age = user.pop("age")

# Delete a key permanently
del user["city"]

# Empty the entire dictionary
user.clear()

5. Iterating Through a Dictionary

By default, looping over a dictionary iterates through its keys. You can target values or key-value pairs specifically.

data = {"a": 1, "b": 2, "c": 3}

# Loop through keys
for key in data:
    print(key)

# Loop through values
for value in data.values():
    print(value)

# Loop through both keys and values
for key, value in data.items():
    print(f"Key: {key}, Value: {value}")

Common Built-in Methods
Method	Descriptionkeys()	Returns a view object of all keys
values()	Returns a view object of all values
items()	Returns a view object of all (key, value) tuple pairs
update(dict2)	Merges another dictionary or iterable of key-values into this one
popitem()	Removes and returns the last inserted key-value pair
Quick Cheat Sheet Example
# Initialize a dictionary
employee = {
    "name": "John",
    "role": "Developer"
}

# Add a new key-value pair
employee["department"] = "Engineering"

# Modify a value
employee["role"] = "Senior Developer"

# Access values
print(employee["name"])          # John
print(employee.get("salary"))    # None

# Iterate through dictionary
for key, value in employee.items():
    print(key, value)

# Remove an item
employee.pop("department")

# Check dictionary size
print(len(employee))             # Output: 2
