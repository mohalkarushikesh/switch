Python Lists

A Python list is a built-in, ordered, and mutable data structure used to store collections of items inside square brackets []. Lists are highly versatile because they can hold duplicate values and a mix of different data types simultaneously.

Basic Syntax
# Creating lists

fruits = ["apple", "banana", "cherry"]
mixed_list = [1, "hello", 3.14, True]
empty_list = []

Core Characteristics
Ordered: Items maintain their specific insertion order.
Mutable: Elements can be added, removed, or changed after creation.
Indexed: Items are accessed using zero-based indexing (e.g., fruits[0] is "apple").
Dynamic: They automatically grow or shrink as you modify them.
Essential Operations & Methods
1. Accessing & Slicing Elements

Python supports both positive indexing from the front and negative indexing from the back.

fruits[1] → Returns "banana" (second item).
fruits[-1] → Returns "cherry" (last item).
fruits[0:2] → Returns ["apple", "banana"] (slicing up to, but not including, index 2).
2. Adding Items
append(item) → Adds an element to the exact end of the list.
insert(index, item) → Inserts an item at a specific position.
extend(iterable) → Appends all items from another collection to the list.
3. Removing Items
remove(item) → Removes the very first occurrence of a specific value.
pop(index) → Removes and returns the item at a given index (defaults to the last item).
clear() → Removes every single item, leaving the list completely empty.
4. Utility Functions
len(list) → Returns the total number of elements.
sort() → Sorts the list items in-place.
reverse() → Reverses the order of the list items in-place.
Common Cheat Sheet Example
# Initialize a list
tech_stack = ["Python", "SQL"]

# 1. Add elements
tech_stack.append("Git")          # ['Python', 'SQL', 'Git']
tech_stack.insert(1, "Docker")    # ['Python', 'Docker', 'SQL', 'Git']

# 2. Modify elements
tech_stack[2] = "PostgreSQL"      # ['Python', 'Docker', 'PostgreSQL', 'Git']

# 3. Remove elements
tech_stack.remove("Docker")       # ['Python', 'PostgreSQL', 'Git']
last_item = tech_stack.pop()      # Removes 'Git'

# 4. Check length
print(len(tech_stack))            # Output: 2
