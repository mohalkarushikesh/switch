# Python Tuples

A **tuple** is an ordered, *immutable* collection of items. Once created, it cannot be changed, resized, or reordered — which makes tuples memory-efficient, hashable, and safe for fixed data.

**Topics covered in this notebook**

1. Creating a tuple
2. Accessing elements (indexing & slicing)
3. Unpacking a tuple
4. Immutability & the mutable-object exception
5. Tuples as dictionary keys
6. Tuple methods — `count()` and `index()`
7. Membership testing & length

**1. Creating a Tuple**

Tuples are created with parentheses `()` and comma-separated values. A **single-element** tuple needs a trailing comma — without it, Python treats the value as its plain type (e.g. a string), not a tuple.

```python
empty_tuple = ()
```

```python
empty_tuple
```

```python
# tuples with mixed data types 
mixed_data_types = ("krishna", 84, "Bharat")
```

```python
mixed_data_types
```

```python
# CRITICAL: Single-element tuples require a trailing comma
fruits_tuple = ("Apple")    # Without the comma, Python sees a string
```

```python
type(fruits_tuple)
```

```python
fruits_tupl = ("Banana", )
fruits_tupl
```

```python
type(fruits_tuple)
```

```python
fruits = ("Apple", "Banana", "Cheery")
```

**2. Accessing Elements**

Use square brackets `[]` with a zero-based index. Negative indices count from the end (`-1` is the last item), and slicing `[start:stop]` returns a sub-tuple.

```python
print(fruits[0])
print(fruits[-1])
print(fruits[1: 3])
```

**3. Unpacking a Tuple**

Tuple unpacking assigns each element to its own variable in a single statement — commonly used when a function returns multiple values.

```python
directions = ("east", "west", "south")
```

```python
e, w, s = directions

print(e)
print(w)
print(s)
```

**4. Immutability Exceptions**

A tuple itself is immutable, but if it contains a **mutable** object (like a list), that nested object can still be modified.

```python
my_tuple = (1, 2, [4, 5])
my_tuple[2][0] = 99

print(my_tuple)
```

**5. Tuples as Dictionary Keys**

Because tuples are immutable and **hashable**, they can serve as dictionary keys. Lists cannot — they are mutable and unhashable.

```python
# Valid use case
locations = {
    (40.7128, -74.0060): "New York",
    (34.0522, -118.2437): "Los Angeles"
}
```

```python
locations
```

**6. Tuple Methods — `count()` and `index()`**

Tuples have only two built-in methods: `count()` returns how many times a value appears, and `index()` returns the first position of a value.

```python
numbers = (1, 2, 2, 3, 2)
numbers.count(2)   # how many times 2 appears -> 3
```

```python
nums = (10, 20, 30, 20)
nums.index(20)   # first index where 20 appears -> 1
```

**7. Membership & Length**

Use the `in` operator to test whether a value exists in a tuple, and `len()` to count its elements.

```python
person = ("Rushikesh", 25, "AIML Engineer")

print("Rushikesh" in person)   # membership test -> True
print(len(person))             # number of elements -> 3
```
