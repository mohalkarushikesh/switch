# Python Functions

A **function** is a reusable, named block of code that takes inputs, performs a task, and optionally returns a result. Python offers flexible ways to define parameters and return values.

**Topics covered in this notebook**

1. Default Arguments
2. Variable-Length Arguments (`*args` and `**kwargs`)
3. Lambda Functions
4. Returning Multiple Values

**1. Default Arguments**

Default arguments give parameters a fallback value used when no argument is passed for them. Parameters without defaults must always come before those with defaults.

```python
# default arguments - fallback values when no value is passed
def greet(name, message="welcome"):
    return f"{message}, {name}"
```

```python
print(greet("Rushikesh"))                 # uses default message
```

```python
print(greet("Rushikesh", "Good morning")) # overrides default
```

**2. Variable-Length Arguments (`*args` and `**kwargs`)**

`*args` packs an arbitrary number of extra positional arguments into a tuple, while `**kwargs` packs named keyword arguments into a dictionary. Use them when you don't know how many inputs a function will receive.

```python
# *args packs extra positional arguments into a tuple
def multiply_all(*args):
    result = 1 
    for num in args:
        result *= num
    return result 
```

```python
print(multiply_all(2, 3, 4))
```

```python
# **kwargs packs named keyword arguments into a dict
def profile(**kargs):
    for key, value in kargs.items():
        print(f"{key}, {value}")
```

```python
print(profile(name= "Rushikesh", age= 25, goal="top 1%"))
```

**3. Lambda Functions**

Lambda functions are small, anonymous functions defined without the `def` keyword. They can take any number of arguments but are limited to a single expression that is returned automatically.

```python
# small anonymous single-expression function
square = lambda x: x ** 2
square(5)
```

```python
# lambda as a sort key
pairs = [(1, 'one'), (2, 'two'), (3, 'three')]
pairs.sort(key=lambda pair: pair[1])  # sort alphabetically by the word
pairs
```

**4. Returning Multiple Values**

A function can return several values by separating them with commas in the `return` statement. Python packs them into a single tuple, which can then be unpacked into separate variables.

```python
# comma-separated return values are packed into a tuple
def get_stats(numbers):
    total = sum(numbers)
    count = len(numbers)
    return total, count

# unpack into separate variables
my_sum, my_count = get_stats([10, 20, 30])
print("Sum:", my_sum)
print("Count:", my_count)
```
