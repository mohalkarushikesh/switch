**1. Iterators & Generators (`yield`)**

A generator produces values lazily — one at a time, only when asked — instead of building the whole collection in memory. Use a function with `yield`, or a generator expression `(...)`. This matters for streaming large data without blowing up memory.

```python
# generator function - 'yield' produces one value at a time, pausing in between
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

gen = count_up_to(5)
print(gen)          # a generator object (nothing computed yet)
print(list(gen))    # consume it -> [1, 2, 3, 4, 5]
```

```python
# generator expression - like a list comprehension but lazy, using ()
squares = (x ** 2 for x in range(5))

print(next(squares))  # 0
print(next(squares))  # 1
print(list(squares))  # remaining values: [4, 9, 16]
```

**2. `enumerate`, `zip`, `map`, `filter`**

These built-ins replace clunky index bookkeeping: `enumerate` yields `(index, value)` pairs, `zip` walks several iterables together, `map` applies a function to every item, and `filter` keeps only items that pass a test.

The enumerate() function in Python is a built-in tool that allows you to loop through an iterable (like a list, tuple, or string) while automatically keeping track of the index of each item. It eliminates the need to manually initialize and increment a counter variable outside your loop

```python
# enumerate - get the index alongside each value while looping
names = ["Rushikesh", "Arjun", "Karan"]

for i, name in enumerate(names, start=1):
    print(i, name)
```

```python
# zip - pair up items from multiple iterables in lockstep
names = ["Rushikesh", "Arjun", "Karan"]
scores = [90, 85, 95]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```

```python
# map - apply a function to every item; filter - keep items matching a condition
nums = [1, 2, 3, 4, 5, 6]

doubled = list(map(lambda x: x * 2, nums))
evens = list(filter(lambda x: x % 2 == 0, nums))

print("doubled:", doubled)
print("evens:  ", evens)
```

**3. Decorators**

A decorator wraps a function or method to add behavior without changing its body, applied with `@name` above the definition. Common built-ins: `@property` (access a method like an attribute) and `@staticmethod`. Frameworks use them too (e.g. FastAPI's `@app.get`).

```python
# @property - call a method like an attribute (no parentheses)
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def area(self):
        return 3.14159 * self.radius ** 2

c = Circle(5)
c.area   # accessed as c.area, not c.area()
```

```python
# @staticmethod - a method that needs neither 'self' nor 'cls'
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

MathUtils.add(3, 4)
```

```python
# writing your own decorator - wraps a function to add behavior
import functools

def shout(func):
    @functools.wraps(func)          # preserves the original function's name/docstring
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return wrapper

@shout
def greet(name):
    return f"hello, {name}"

greet("rushikesh")
```

**4. Context Managers**

A context manager (`with ...:`) guarantees setup and cleanup around a block — like closing a file even if an error is raised. You can write your own using the `@contextmanager` decorator from `contextlib`.

```python
# 'with' handles setup/teardown automatically - the file is closed for you
with open("demo.txt", "w") as f:
    f.write("Hello buddy!")

with open("demo.txt") as f:
    print(f.read())
```

```python
# write your own context manager with contextlib
from contextlib import contextmanager

@contextmanager
def tag(name):
    print(f"<{name}>")   # runs on entering the 'with' block
    yield
    print(f"</{name}>")  # runs on exiting, even if an error occurs

with tag("p"):
    print("hello")
```

**5. Type Hints**

Type hints annotate the expected types of parameters and return values, e.g. `def embed(text: str) -> list[float]:`. They aren't enforced at runtime, but they document intent and power editor autocomplete, type checkers, and reviews.

```python
# type hints document the expected input/output types
# (they are NOT enforced at runtime, but tools and reviewers rely on them)
def embed(text: str) -> list[float]:
    # pretend this returns an embedding vector
    return [float(len(text))]

vector: list[float] = embed("hello")
vector
```
