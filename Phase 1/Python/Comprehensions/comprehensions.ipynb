# Python Comprehensions and Generator Expressions

Comprehensions and generator expressions provide a concise, readable way to build or transform collections from an existing iterable in a single line — generators do this lazily, one item at a time.

**Topics covered in this notebook**

1. List Comprehensions
2. Set Comprehensions
3. Dictionary Comprehensions
4. Generator Expressions
5. Conditional Logic (Inline Ternary)

**Core Syntax Comparison**

| Type | Enclosure | Syntax Pattern | Resulting Object |
| --- | --- | --- | --- |
| List Comprehension | `[]` | `[expr for item in iterable if cond]` | A new list object |
| Set Comprehension | `{}` | `{expr for item in iterable if cond}` | A new set (unique values) |
| Dictionary Comprehension | `{}` | `{key: value for item in iterable if cond}` | A new dict object |
| Generator Expression | `()` | `(expr for item in iterable if cond)` | A lazy generator iterator |

**1. List Comprehensions**

A list comprehension builds a new list by applying an expression to each item of an iterable, in a single readable line that replaces a `for`-append loop.

```python
# traditional loop 
squares = []

for x in range(5):
    squares.append(x**2)
    
squares 
```

```python
# comprehensions 

squares_2 = [x**2 for x in range(10)]
squares_2
```

**Adding an `if` filter**

Placing an `if` clause at the end keeps only the items for which the condition is true.

```python
even_nums = [x for x in range(10) if x % 2 == 0]
even_nums 
```

**2. Set Comprehensions**

A set comprehension uses curly braces `{}` to build a set, automatically discarding duplicate values.

```python
words = ["apple", "banana", "APPLE", "BANANA"]

unique_words = {w.lower() for w  in words}

unique_words
```

**3. Dictionary Comprehensions**

A dictionary comprehension builds a dict by producing a `key: value` pair for each item in an iterable.

```python
names = ["Rushikesh", "Arjun", "Mohalkar"]

name_with_len = {name: len(name) for name in names}

count = sum(name_with_len.values())
print("Total chars: ", count)

name_with_len
```

**4. Generator Expressions**

Generator expressions use parentheses `()` and evaluate lazily — instead of storing the whole collection in memory, they hold the formula and produce one item at a time on demand.

```python
# Does not computes values immediately 

lazy_squares = (x**2 for x in range(100000))

print(lazy_squares)                
                
```

```python
# Values are evaluated on demand (e.g., using next() or a loop)

next(lazy_squares)
```

```python
next(lazy_squares)
```

**5. Conditional Logic (Inline Ternary)**

To transform items based on a condition, place a ternary `expr_if_true if condition else expr_if_false` at the front of the comprehension (unlike a filtering `if`, which goes at the end).

```python
# Syntax: [expr_if_true if condition else expr_if_false for item in iterable]

labels = ["even" if x % 2 == 0 else "odd" for x in range(10)]

labels
```
