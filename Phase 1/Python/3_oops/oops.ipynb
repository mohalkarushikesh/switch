# Object-Oriented Programming (OOP)

Classes bundle data (attributes) and behavior (methods) into reusable objects. Grasping them makes library code click — e.g. wrapping text chunks as LangChain `Document` objects in a RAG app.

**Topics covered in this notebook**

1. Classes, `__init__` & methods
2. Inheritance
3. `__str__` and `__repr__`
4. Dunder (magic) methods

**1. Classes, `__init__` & Methods**

A class is a blueprint for objects. `__init__` is the constructor that runs when you create an instance, setting up its attributes; methods are functions defined inside the class, and `self` refers to the current instance.

```python
# a class bundles data (attributes) + behavior (methods)
class Dog:
    def __init__(self, name, breed):   # constructor - runs on object creation
        self.name = name               # instance attributes
        self.breed = breed

    def bark(self):                    # method - 'self' is the instance
        return f"{self.name} says woof!"

d = Dog("Bruno", "Labrador")
print(d.name, "-", d.breed)
d.bark()
```

**2. Inheritance**

A subclass inherits the attributes and methods of a parent class and can add or override behavior. `super()` calls the parent's version of a method — commonly used in `__init__` to reuse the parent's setup.

```python
# inheritance - a subclass reuses and extends a parent class
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "..."

class Cat(Animal):                     # Cat inherits from Animal
    def __init__(self, name, indoor):
        super().__init__(name)         # call the parent's __init__
        self.indoor = indoor

    def speak(self):                   # override the parent method
        return f"{self.name} says meow!"

c = Cat("Whiskers", indoor=True)
print(c.name, "| indoor:", c.indoor)   # 'name' came from Animal
print(c.speak())                       # overridden version
```

**3. `__str__` and `__repr__`**

Both control how an object is shown as text. `__str__` is the friendly version used by `print()`; `__repr__` is the unambiguous, developer-facing version shown in the REPL and debuggers. If you define only one, define `__repr__`.

```python
# __str__  -> friendly text for print()/str()
# __repr__ -> unambiguous text for developers/debugging (shown in the REPL)
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Point(x={self.x}, y={self.y})"

p = Point(3, 4)
print(p)      # uses __str__  -> (3, 4)
p             # uses __repr__ -> Point(x=3, y=4)
```

**4. Dunder (Magic) Methods**

Dunder methods (double-underscore, like `__add__`, `__eq__`, `__len__`) hook your objects into Python's built-in operators and functions. Defining them lets you write `v1 + v2`, `a == b`, or `len(obj)` on your own classes — this is how library code stays readable.

```python
# dunder ("double underscore") methods let your objects work with
# built-in operators and functions
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):          # enables  v1 + v2
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):           # enables  v1 == v2
        return (self.x, self.y) == (other.x, other.y)

    def __len__(self):                 # enables  len(v)
        return 2

v1, v2 = Vector(1, 2), Vector(3, 4)
print(v1 + v2)                 # __add__
print(v1 == Vector(1, 2))      # __eq__
print(len(v1))                 # __len__
```
