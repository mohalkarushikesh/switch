**1. Virtual Environments & Packaging**

A virtual environment isolates a project's dependencies so they don't clash across projects. `pip` installs packages, and `requirements.txt` pins exact versions so others can reproduce your setup. These are **terminal** commands, not Python:

```bash
python -m venv .venv               # create an isolated environment
source .venv/bin/activate          # activate it  (Windows: .venv\Scripts\activate)

pip install requests               # install a package
pip freeze > requirements.txt      # save exact versions
pip install -r requirements.txt    # reproduce the environment elsewhere
```

**2. Error Handling**

`try`/`except` catches errors so your program doesn't crash; `else` runs when no error occurred, and `finally` always runs (cleanup). You can `raise` built-in exceptions or define your own by subclassing `Exception`.

````python
# try / except / else / finally
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
        return None
    else:
        return result          # runs only if no exception was raised
    finally:
        print("(division attempt finished)")   # always runs

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

````python
# raising and catching your own custom exception
class InvalidChunkError(Exception):
    """Raised when a text chunk is empty."""

def make_chunk(text):
    if not text.strip():
        raise InvalidChunkError("chunk text is empty")
    return text

try:
    make_chunk("   ")
except InvalidChunkError as e:
    print("Caught:", e)
````

**3. Files, JSON & Paths (`pathlib`)**

`json.dumps`/`json.loads` convert between Python objects and JSON strings, while `pathlib.Path` gives an object-oriented, cross-platform way to read, write, and inspect file paths (`.read_text()`, `.write_text()`, `.suffix`, `.exists()`).

````python
import json
from pathlib import Path

# --- JSON: dict <-> string <-> file ---
data = {"model": "phi-1.5", "chunks": 42}
path = Path("config.json")

path.write_text(json.dumps(data))        # dict -> JSON string -> file
loaded = json.loads(path.read_text())    # file -> JSON string -> dict
print(loaded)

# --- pathlib: clean, cross-platform path handling ---
print("name:  ", path.name)     # config.json
print("suffix:", path.suffix)   # .json
print("stem:  ", path.stem)     # config
print("exists:", path.exists()) # True

path.unlink()   # clean up the demo file
````

**4. f-strings**

f-strings (`f"..."`) embed expressions directly inside string literals using `{}`. They support format specifiers for decimals (`:.2f`), percentages (`:.1%`), alignment (`:<10`), `repr` (`!r`), and a handy debug form (`{value=}`).

````python
name, score = "Rushikesh", 0.9137

print(f"{name} scored {score}")            # basic interpolation
print(f"{score:.2f}")                      # 2 decimal places  -> 0.91
print(f"{score:.1%}")                      # as a percentage   -> 91.4%
print(f"{name!r}")                         # repr()            -> 'Rushikesh'
print(f"{'left':<10}|{'right':>10}")       # left / right alignment
print(f"{score=}")                         # debug form        -> score=0.9137
````
