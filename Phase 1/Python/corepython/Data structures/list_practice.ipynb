# Python Lists

A **list** is an ordered, *mutable* collection of items stored in square brackets `[]`. Lists can hold duplicate values and a mix of data types, and they grow or shrink dynamically as items are added or removed.

**Topics covered in this notebook**

1. Creating a list
2. Accessing & slicing elements
3. Adding elements
4. Removing items
5. Length
6. Sorting & reversing (in-place)

**1. Creating a List**

Lists are written as comma-separated values inside square brackets `[]`. They can mix data types and contain duplicate values.

```python
fruits = ['apple', 'banana', 'grapes', 'guava']
```

**2. Accessing & Slicing Elements**

Items are retrieved by zero-based index inside square brackets `[]`. Negative indices count from the end (`-1` is the last item), and slicing `[start:stop]` returns a sub-list up to (but not including) the stop index.

**1. Accessing & Slicing Elements**

Items are retrieved by zero-based index inside square brackets `[]`. Negative indices count from the end (`-1` is the last item), and slicing `[start:stop]` returns a sub-list up to (but not including) the stop index.

```python
# index based access 
fruits[0]
```

```python
# returns the last item
fruits[-1]
```

**3. Adding Elements**

`append()` adds a single item to the end of the list, `insert()` places an item at a specific index, and `extend()` appends all items from another iterable.

**2. Adding Elements**

`append()` adds a single item to the end of the list, `insert()` places an item at a specific index, and `extend()` appends all items from another iterable.

```python
fruits.append("Pineapple")
```

**4. Removing Items**

`remove()` deletes the first matching value, `pop()` removes and returns the item at a given index (the last item by default), and `clear()` empties the list entirely.

**3. Removing Items**

`remove()` deletes the first matching value, `pop()` removes and returns the item at a given index (the last item by default), and `clear()` empties the list entirely.

```python
fruits.remove("cherry")
```

```python
fruits
```

```python
# removes and returns the items at given index 
fruits.pop(2)
```

```python
fruits
```

```python
# removes every single item
fruits.clear()
```

**5. Length**

`len()` returns the number of items currently in the list.

```python
fruits
```

**6. Sorting & Reversing (In-Place)**

In-place operations rearrange the elements directly within the list's original memory location without creating a new copy. `sort()` orders the items, and `reverse()` flips their current order.

```python
len(fruits)
```

**4. Sorting & Reversing (In-Place)**

In-place operations rearrange the elements directly within the list's original memory location without creating a new copy. `sort()` orders the items, and `reverse()` flips their current order.
