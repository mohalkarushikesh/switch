# Slicing & Unpacking

**Extended unpacking** (starred assignment) uses a starred variable such as `a, *rest = mylist` to assign some items to individual variables while collecting the remaining items into a new list.

**Topics covered in this notebook**

1. Extended unpacking (starred assignment)

**1. Extended Unpacking (Starred Assignment)**

`a, *rest = mylist` assigns the first item to `a` and collects everything else into a new list `rest` (an empty list if there's nothing left). Rules: only one `*` is allowed per assignment; the starred variable can sit in any position (e.g. `*rest, last = mylist`); and it works on any iterable — tuples, strings, and more.

```python
fruits = ["apple", "banana", "pineapple"]
```

```python
a, *rest = fruits 
```

```python
a 
```

```python
rest 
```
