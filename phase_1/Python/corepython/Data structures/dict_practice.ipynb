# Python Dictionaries

A **dictionary** (`dict`) is a mutable, ordered collection that stores data as **key-value pairs**, giving fast lookups by key. Keys must be unique and immutable (e.g. strings, numbers, tuples); values can be any type and may repeat.

**Topics covered in this notebook**

1. Creating a dictionary
2. Accessing values
3. Adding and modifying items
4. Removing items
5. Iterating through a dictionary
6. Common built-in methods

**1. Creating a Dictionary**

Define a dictionary with curly braces `{}` holding `key: value` pairs, or build one with the `dict()` constructor using keyword arguments.

```python
# using curly braces 
users = {
    "name" : "Rushikesh",
    "age" : 25,
    "goal" : "top 1 percent!"
}
```

```python
users
```

```python
# using dict constructor
user = dict(name="Rushikesh", age=25, goal="to become top 1 %!")
```

```python
user 
```

**2. Accessing Values**

Fetch a value with square brackets `dict[key]`, which raises a `KeyError` if the key is missing. The `.get()` method is safer — it returns `None` (or a supplied default) instead of raising.

```python
# using the square brackets (Raises a keyerror if the key is missing)
# print(user["name3"])
```

```python
print(user["name"])
```

```python
# Using .get() (Returns None or a default value if key is missing)
user.get("name3")
```

```python
user.get("name")
```

```python
print(users.get("email", "N/A"))
```

**3. Adding and Modifying Items**

Assign with `dict[key] = value`: if the key already exists its value is updated, otherwise a new key-value pair is added.

```python
# update the value 

user["age"] = 27
```

```python
print(user.get("age"))
```

```python
# add new key value pair 

user["email"] = "rushi@ex.com"
```

```python
user
```

**4. Removing Items**

Use `.pop(key)` to remove a key and return its value, `del dict[key]` to delete a key permanently, or `.clear()` to empty the dictionary entirely.

```python
# remove the specific key and returns it's value 

age = user.pop("age")
```

```python
print(age)
```

```python
user
```

```python
# delete the key permanently 

del user["email"]
```

```python
user
```

```python
# clear the entire dict 

user.clear()
```

```python
user
```

**5. Iterating Through a Dictionary**

Looping over a dictionary iterates its keys by default. Use `.values()` to loop over values and `.items()` to loop over `(key, value)` pairs together.

```python
users
```

```python
# loop through the keys 

for key in users:
    print(key)
```

```python
for value in users:
    print(value)
```

```python
for key, value in users.items():
    print(f"key: {key}, value: {value}")
```

**6. Common Built-in Methods**

`keys()`, `values()`, and `items()` return view objects of the keys, values, and pairs. `update()` merges another dictionary in, and `popitem()` removes and returns the last inserted pair.

```python
users.keys()
```

```python
users.values()
```

```python
users.items()
```

```python
users2 = {
    "lastname": "Mohalkar", 
    "hobby": "avid reader",
    "city": "Bangalore"
}
```

```python
# Merges another dictionary or iterable of key-values into this one
users.update(users2)
```

```python
users
```

```python
# Removes and returns the last inserted key-value pair

users2.popitem()
```

```python
users2.update(city="Pune")
users2
```

```python
users2["role"]="AIML Engineer"
users
```
