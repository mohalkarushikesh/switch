# Python Sets

A **set** is an unordered collection of *unique*, immutable items. Sets automatically discard duplicate values and are highly optimized for fast membership testing.

**Topics covered in this notebook**

1. Creating a set
2. Adding & removing elements
3. Checking for membership
4. Mathematical set operations — union, intersection, difference & symmetric difference

**1. Creating a Set**

Sets are created with curly braces `{}` or the `set()` constructor. Use `set()` for an empty set — empty `{}` creates a dictionary, not a set.

```python
# with curly braces 
fruits = {"apple", "banana", "lemon"}
```

```python
fruits 
```

```python
# using set contrustor 

empty_set = set()
```

```python
new_set_of_fruits = set(["pineapple", "cherry"])
```

**2. Adding & Removing Elements**

Use `add()` for a single item and `update()` to add multiple items from an iterable. `remove()` deletes an item but raises `KeyError` if it is missing, while `discard()` removes safely without error.

```python
numbers = {1, 2, 3}
```

```python
numbers.add(4)
```

```python
numbers 
```

```python
# Add mulitple items from another iterable 
numbers.update([5, 6])
```

```python
numbers 
```

```python
numbers.remove(2)
```

```python
numbers 
```

```python
# Remove an item safely (does nothing if not found)
numbers.discard(4)
```

```python
numbers
```

**3. Checking for Membership**

Use the `in` operator to test whether a value exists in a set. Thanks to hashing, this check runs in O(1) average time.

```python
print(5 in numbers)
```

**4. Mathematical Set Operations**

Sets support mathematical operations: `union` (`|`) combines all unique items, `intersection` (`&`) keeps shared items, `difference` (`-`) keeps items only in the first set, and `symmetric_difference` (`^`) keeps items unique to each set.

```python
# Union - combines all the common items from both sets 
# operator : set1 | set2 

set1 = {1, 2, 3}
set2 = {3, 4, 5}

set_onion = set1.union(set2)
set_onion 
```

```python
# Intersection - keeps the items present in both tables 
# operator : set1 and set2 

set_itrsec = set1.intersection(set2)
set_itrsec
```

```python
# difference - keeps the items in set1 that are not in set2 
# operator = set1 - set2

set_diff = set1.difference(set2)
set_diff
```

```python
# symmetric diff - kep items unique to each set (exclude common items)
# operator = set1 ^ set2

set_symm_diff = set1.symmetric_difference(set2)
set_symm_diff
```
