# Python Data Structures — Cheat Sheet

Quick reference for the four core built-in collections.

## At a Glance

| Type | Syntax | Ordered | Mutable | Duplicates | Access by |
| --- | --- | --- | --- | --- | --- |
| **List** | `[1, 2, 3]` | ✅ | ✅ | ✅ | index |
| **Dict** | `{"k": "v"}` | ✅ | ✅ | keys unique, values may repeat | key |
| **Set** | `{1, 2, 3}` / `set()` | ❌ | ✅ | ❌ (auto-deduped) | membership only |
| **Tuple** | `(1, 2, 3)` | ✅ | ❌ | ✅ | index |

> Lists and dicts preserve insertion order (Python 3.7+). Sets have no order and no indexing. Tuples are hashable — usable as dict keys / set members — as long as all their elements are.

## List — `[]`

```python
nums = [1, 2, 3]

# access & slice
nums[0]; nums[-1]; nums[1:3]

# add
nums.append(4)           # add to end
nums.insert(1, 99)       # add at index
nums.extend([5, 6])      # merge another iterable

# remove
nums.remove(99)          # first matching value
nums.pop()               # last item (or pop(i))
nums.clear()             # empty the list

# order & size
nums.sort(); nums.reverse()   # in-place
len(nums)
```

## Dictionary — `{key: value}`

```python
user = {"name": "Alice", "age": 25}

# access
user["name"]             # KeyError if key missing
user.get("email", "N/A") # safe access with a default

# add / modify
user["age"] = 26         # update existing key
user["email"] = "a@x.com"  # add new key

# remove
user.pop("age")          # remove & return value
del user["email"]        # delete a key
user.clear()             # empty the dict

# iterate & views
for k, v in user.items():
    print(k, v)
user.keys(); user.values(); user.items()
user.update({"city": "Pune"})  # merge another dict
user.popitem()           # remove & return last inserted pair
```

## Set — `{}` / `set()`

```python
s = {1, 2, 3}
empty = set()            # NOTE: {} creates an empty DICT, not a set

# add / remove
s.add(4)
s.update([5, 6])         # add many
s.remove(6)              # KeyError if missing
s.discard(99)            # safe - no error if missing

# membership
3 in s

# math operations
a, b = {1, 2, 3}, {2, 3, 4}
a | b                    # union            -> {1, 2, 3, 4}
a & b                    # intersection     -> {2, 3}
a - b                    # difference       -> {1}
a ^ b                    # symmetric diff   -> {1, 4}
```

## Tuple — `()`

```python
point = (4, 10)
single = ("x",)          # trailing comma REQUIRED for a one-item tuple

# access & slice
point[0]; point[-1]; point[0:2]

# unpack
x, y = point
first, *rest = (1, 2, 3, 4)   # extended unpacking -> first=1, rest=[2,3,4]

# methods (tuples have only two)
nums = (1, 2, 2, 3)
nums.count(2)            # -> 2   occurrences of a value
nums.index(3)            # -> 3   first position of a value

# immutable but hashable -> valid dict key
locations = {(40.71, -74.0): "NYC"}
```
